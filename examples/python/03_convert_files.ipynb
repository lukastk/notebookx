# Converting Notebook Files with notebookx

This notebook demonstrates how to convert notebook files using `notebookx`:

- Using the `convert()` convenience function
- Using the `clean_notebook()` convenience function
- File format inference from extensions

## Setup

First, let's import the necessary modules:

```python
import os
import tempfile
from notebookx import convert, clean_notebook, Notebook, Format
```

## Creating a Temporary Directory

We'll use a temporary directory to store our converted files:

```python
# Create a temporary directory for output files
tmpdir = tempfile.mkdtemp()
print(f"Working in: {tmpdir}")
```

## Converting ipynb to Percent Format

The `convert()` function provides a simple way to convert between formats. The input and output formats are automatically inferred from the file extensions.

```python
# Convert ipynb to percent format
input_file = "../../nb_format_examples/World population.ipynb"
percent_output = os.path.join(tmpdir, "notebook.pct.py")

convert(input_file, percent_output)
print(f"Converted {input_file} -> {percent_output}")
```

## Inspecting the Percent Format

Let's look at the first few lines of the converted file:

```python
# Read and show first few lines
with open(percent_output) as f:
    content = f.read()
print("Percent format preview (first 300 chars):")
print(content[:300])
```

## Converting Back to ipynb

We can convert the percent file back to ipynb format:

```python
# Convert back to ipynb
ipynb_output = os.path.join(tmpdir, "restored.ipynb")
convert(percent_output, ipynb_output)
print(f"Converted back: {percent_output} -> {ipynb_output}")
```

## Verifying the Round-Trip

Let's verify that the cell count is preserved through the round-trip conversion:

```python
# Verify round-trip
original = Notebook.from_file(input_file)
restored = Notebook.from_file(ipynb_output)
print(f"Cell count: {len(original)} -> {len(restored)}")
```

## Using the clean_notebook() Function

The `clean_notebook()` convenience function makes it easy to clean a notebook file directly:

```python
# Clean a notebook to a new file
clean_output = os.path.join(tmpdir, "clean.ipynb")
clean_notebook(input_file, clean_output, remove_outputs=True)
print(f"Cleaned notebook saved to: {clean_output}")
```

## Verifying the Cleaning

Let's verify that the outputs were removed:

```python
import json

# Verify cleaning
clean_nb = Notebook.from_file(clean_output)
clean_json = json.loads(clean_nb.to_string(Format.Ipynb))
code_cells = [c for c in clean_json["cells"] if c["cell_type"] == "code"]
has_outputs = any(c.get("outputs") for c in code_cells)
print(f"Clean notebook has outputs: {has_outputs}")
```

## Cleanup

Let's clean up the temporary directory:

```python
import shutil
shutil.rmtree(tmpdir)
print("Done! Temporary files cleaned up.")
```

## Summary

In this notebook, we've learned how to:

1. Use `convert()` to convert between notebook formats
2. Format is automatically inferred from file extensions (`.ipynb`, `.pct.py`)
3. Use `clean_notebook()` to clean a notebook file directly
4. Verify that conversions and cleaning work correctly

These convenience functions are perfect for scripting and automation!
