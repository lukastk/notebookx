# Basic Usage of notebookx

This notebook demonstrates the basic usage of the `notebookx` Python library:

- Loading notebooks from files
- Converting between formats
- Basic notebook inspection

## Importing the Library

First, let's import the necessary classes from `notebookx`:

```python
from notebookx import Notebook, Format
```

## Loading a Notebook

We can load a notebook from a file using `Notebook.from_file()`. The format is automatically inferred from the file extension.

```python
# Load a notebook from file
nb = Notebook.from_file("../../nb_format_examples/World population.ipynb")
```

## Inspecting the Notebook

The `Notebook` object provides several properties to inspect its contents:

```python
# Inspect the notebook
print(f"Loaded notebook with {len(nb)} cells")
print(f"  - Code cells: {nb.code_cell_count}")
print(f"  - Markdown cells: {nb.markdown_cell_count}")
print(f"  - Raw cells: {nb.raw_cell_count}")
print(f"  - nbformat: {nb.nbformat}.{nb.nbformat_minor}")
```

## Converting to Percent Format

We can convert the notebook to "percent format" (used by Jupytext, VSCode, etc.) using `to_string()` with `Format.Percent`:

```python
# Convert to percent format
percent_content = nb.to_string(Format.Percent)
print("--- Percent format (first 500 chars) ---")
print(percent_content[:500])
```

## Round-Trip Conversion

We can convert the percent format back to a Notebook object and verify the cell count is preserved:

```python
# Convert back from percent format
nb2 = Notebook.from_string(percent_content, Format.Percent)
print(f"Round-trip: {len(nb)} cells -> {len(nb2)} cells")
```

## Summary

In this notebook, we've learned how to:

1. Load a notebook from a file with `Notebook.from_file()`
2. Inspect notebook properties like cell counts and nbformat version
3. Convert to different formats with `to_string(Format.Percent)`
4. Parse a string back to a Notebook with `Notebook.from_string()`
